```python
#Write a word counter: given "the cat saw the dog and the bird", count how many times each word appears and store the counts in a dictionary.
 #(Hint: .split() turns a sentence into a list, and counts.get(word, 0) is handy.)

sentence = "The cat saw the dog. The dog ran!"

words = sentence.split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
```

```python
# Fix it using .lower() and .replace()

sentence = "The cat saw the dog. The dog ran!"

# Convert to lowercase and remove punctuation
sentence = sentence.lower()
sentence = sentence.replace(".", "")
sentence = sentence.replace("!", "")

words = sentence.split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print(word_count)
```

```python
# Counting Twist


tickets = {
    "T101": "Open",
    "T102": "Closed",
    "T103": "open",
    "T104": "Closed",
    "T105": "Open"
}

count = 0

for ticket_id, status in tickets.items():
    if status.lower() == "open":
        print(f"{ticket_id} is {status}")
        count += 1

print(f"{count} tickets are Open")
```

```python
# Given [12, 5, 30, 8, 24], create a new list containing only values over 8. Use a loop and .append().

numbers = [12, 5, 30, 8, 24]

# Create an empty list
filtered_numbers = []

# Keep only values over 8
for num in numbers:
    if num > 8:
        filtered_numbers.append(num)

print(filtered_numbers)
```

```python
# Build a second list with values at or above 8

numbers = [12, 5, 30, 8, 24]

filtered_numbers = []

for num in numbers:
    if num >= 8:
        filtered_numbers.append(num)

print(filtered_numbers)
```

```python
# Find the largest number in a list without using max(). Use a loop and a variable that remembers the biggest so far.

numbers = [12, 5, 30, 8, 24]

# Start by assuming the first number is the largest
largest = numbers[0]

# Check each number in the list
for num in numbers:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

```python
# Build: ticket summary script
# Make a list of 8 tickets, each a dictionary with id, status (“Open” or “Closed”) and hours_open. The script should print:

# Day 1 priority rules
def ticket_priority(hours_open):
    if hours_open >= 30:
        return "Critical"
    elif hours_open >= 9:
        return "High"
    else:
        return "Normal"


tickets = [
    {"id": "T101", "status": "Open",   "hours_open": 30},
    {"id": "T102", "status": "Closed", "hours_open": 5},
    {"id": "T103", "status": "Open",   "hours_open": 9},
    {"id": "T104", "status": "Open",   "hours_open": 8},
    {"id": "T105", "status": "Closed", "hours_open": 26},
    {"id": "T106", "status": "Open",   "hours_open": 2},
    {"id": "T107", "status": "Open",   "hours_open": 24},
    {"id": "T108", "status": "Closed", "hours_open": 12},
]

open_count = 0
closed_count = 0
oldest_ticket = tickets[0]

for ticket in tickets:
    if ticket["status"] == "Open":
        open_count += 1
    elif ticket["status"] == "Closed":
        closed_count += 1

    if ticket["hours_open"] > oldest_ticket["hours_open"]:
        oldest_ticket = ticket

print("Total tickets:", len(tickets))
print("Open tickets:", open_count)
print("Closed tickets:", closed_count)
print(
    "Ticket with highest hours_open:",
    oldest_ticket["id"],
    "-",
    oldest_ticket["hours_open"],
    "hours"
)

print("\nOpen ticket priorities:")

for ticket in tickets:
    if ticket["status"] == "Open":
        priority = ticket_priority(ticket["hours_open"])
        print(ticket["id"], priority)
```
