```python
score = int(input("Enter your score: "))
def grade(score):
    if score >= 75:
        return "Distinction"
    elif score >= 40:
        return "You've passed!"
    else:
        return "Fail"
print(grade(score))

```

```python
ticket_priority = int(input("How many tickets are there : "))
def num(ticket_priority):
    if ticket_priority >= 24:
        return "Critical!!"
    elif ticket_priority >= 8:
        return "High!"
    else:
        return "Normal"
print(num(ticket_priority))
```
