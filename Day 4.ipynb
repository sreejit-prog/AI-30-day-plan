```python
import csv, random

random.seed(42)   # same "random" data every run
categories = ["Email", "Hosting", "Billing", "Domain", "SSL", "Website Down"]
statuses = ["Open", "Closed", "Pending"]
agents = ["Anu", "Ravi", "Meera", "John", "Sana"]

rows = []
for i in range(1, 201):
    rows.append({
        "ticket_id": f"T{1000 + i}",
        "category": random.choice(categories),
        "status": random.choices(statuses, weights=[3, 6, 1])[0],
        "hours_open": round(random.uniform(0.5, 72), 1),
        "agent": random.choice(agents),
    })

with open("tickets.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=rows[0].keys())
    writer.writeheader()
    writer.writerows(rows)

print("Wrote", len(rows), "rows")
```

```python
#Reuse your Day 3 count_by_status on the loaded rows.
# Then write count_by_field(rows, field) that counts by any column, and run it
# for "category" and "agent".

import csv

with open("tickets.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))


def count_by_field(tickets, field):
    counts = {}

    for ticket in tickets:
        value = ticket[field]
        counts[value] = counts.get(value, 0) + 1

    return counts


# Count by different columns
print("By status:", count_by_field(rows, "status"))
print("By category:", count_by_field(rows, "category"))
print("By agent:", count_by_field(rows, "agent"))
```

```python
import csv


# 1. Load tickets and convert hours_open to float
def load_tickets(path):
    try:
        with open(path, newline="", encoding="utf-8") as f:
            rows = list(csv.DictReader(f))
    except FileNotFoundError:
        print("File not found:", path)
        return []

    for row in rows:
        row["hours_open"] = float(row["hours_open"])

    return rows


# 2. Find the ticket with the highest hours_open
def find_oldest(tickets):
    if not tickets:
        return None

    oldest_ticket = tickets[0]

    for ticket in tickets:
        if ticket["hours_open"] > oldest_ticket["hours_open"]:
            oldest_ticket = ticket

    return oldest_ticket


# 3. Load the actual CSV file
rows = load_tickets("tickets.csv")
print(len(rows))  # Expected: 200

# 4. Test handling a missing file
print(len(load_tickets("nope.csv")))  # Expected: 0

# 5. Find and print the oldest ticket
oldest = find_oldest(rows)

if oldest is not None:
    print("Ticket with the highest hours_open:")
    print(oldest)
else:
    print("No tickets found.")
```

```python
# write urgent.csv.
# It should contain only Open tickets that are High or Critical, with a new priority column:

import csv

def ticket_priority(hours_open):
    if hours_open > 24:
        return "Critical"
    elif hours_open > 8:
        return "High"
    return "Normal"


with open("tickets.csv", newline="", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

urgent = []

for row in rows:
    hours = float(row["hours_open"])

    if row["status"] == "Open":
        priority = ticket_priority(hours)

        if priority in ("High", "Critical"):
            row["priority"] = priority
            urgent.append(row)

fieldnames = list(rows[0].keys()) + ["priority"]

with open("urgent.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(urgent)

print("Urgent tickets written:", len(urgent))
```
