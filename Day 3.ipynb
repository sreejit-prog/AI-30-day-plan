```python
# Write safe_int(text) that returns the integer, or None if the text isn’t a valid whole number. Test it with "42", "abc" and "3.5".

def safe_int(text):
    try:
        return int(text)
    except ValueError:
        return None


# Test the function
print(safe_int("42"))
print(safe_int("abc"))
print(safe_int("3.5"))
```

```python
# Write get_number(prompt) that keeps asking until the user types a valid number, then returns it as a float.

def get_number(prompt):
    while True:
        try:
            number = float(input(prompt))
            return number
        except ValueError:
            print("Invalid input. Please enter a valid number.")


# Test the function
value = get_number("Enter a number: ")
print("Your number is:", value)
```

```python
# Write count_by_status(tickets) that returns a dictionary like {"Open": 5, "Closed": 3}. Use the .get() pattern from Day 2.

def count_by_status(tickets):
    counts = {}

    for ticket in tickets:
        status = ticket["status"]
        counts[status] = counts.get(status, 0) + 1

    return counts


# Test with 8 tickets
tickets = [
    {"id": "T101", "status": "Open"},
    {"id": "T102", "status": "Closed"},
    {"id": "T103", "status": "Open"},
    {"id": "T104", "status": "Open"},
    {"id": "T105", "status": "Closed"},
    {"id": "T106", "status": "Open"},
    {"id": "T107", "status": "Open"},
    {"id": "T108", "status": "Closed"}
]

result = count_by_status(tickets)
print(result)
```

```python
# Write find_oldest(tickets) that returns the ticket with the highest hours_open, or None if the list is empty.

def find_oldest(tickets):
    if not tickets:
        return None

    oldest_ticket = tickets[0]

    for ticket in tickets:
        if ticket["hours_open"] > oldest_ticket["hours_open"]:
            oldest_ticket = ticket

    return oldest_ticket


# Test with tickets
tickets = [
    {"id": "T101", "status": "Open", "hours_open": 30},
    {"id": "T102", "status": "Closed", "hours_open": 5},
    {"id": "T103", "status": "Open", "hours_open": 9},
    {"id": "T104", "status": "Open", "hours_open": 8},
    {"id": "T105", "status": "Closed", "hours_open": 26}
]

print(find_oldest(tickets))
print(find_oldest([]))
```

```python
# Tickets priority handling

# 1. Calculate ticket priority
def ticket_priority(hours_open):
    if hours_open > 24:
        return "Critical"
    elif hours_open > 8:
        return "High"
    else:
        return "Normal"


# 2. Count tickets by status
def count_by_status(tickets):
    counts = {}

    for ticket in tickets:
        status = ticket["status"]
        counts[status] = counts.get(status, 0) + 1

    return counts


# 3. Find the ticket with the highest hours_open
def find_oldest(tickets):
    if not tickets:
        return None

    oldest_ticket = tickets[0]

    for ticket in tickets:
        if ticket["hours_open"] > oldest_ticket["hours_open"]:
            oldest_ticket = ticket

    return oldest_ticket


# 4. Test ticket_priority
assert ticket_priority(25) == "Critical"
assert ticket_priority(24) == "High"
assert ticket_priority(9) == "High"
assert ticket_priority(8) == "Normal"
assert ticket_priority(2) == "Normal"

print("All tests passed")


# 5. Ticket data
tickets = [
    {"id": "T101", "status": "Open",   "hours_open": 30},
    {"id": "T102", "status": "Closed", "hours_open": 5},
    {"id": "T103", "status": "Open",   "hours_open": 9},
    {"id": "T104", "status": "Open",   "hours_open": 8},
    {"id": "T105", "status": "Closed", "hours_open": 26},
    {"id": "T106", "status": "Open",   "hours_open": 2},
    {"id": "T107", "status": "Open",   "hours_open": 24},
    {"id": "T108", "status": "Closed", "hours_open": 12},
]


# 6. Generate the complete report
def print_report(tickets):
    if not tickets:
        print("No tickets to report.")
        return

    status_counts = count_by_status(tickets)
    oldest = find_oldest(tickets)

    print("Total tickets:", len(tickets))
    print("Open tickets:", status_counts.get("Open", 0))
    print("Closed tickets:", status_counts.get("Closed", 0))

    print(
        "Ticket with highest hours_open:",
        oldest["id"],
        "-",
        oldest["hours_open"],
        "hours"
    )

    print("\nOpen ticket priorities:")

    for ticket in tickets:
        if ticket["status"] == "Open":
            priority = ticket_priority(ticket["hours_open"])
            print(ticket["id"], priority)


# 7. Run the report with normal and empty lists
print_report(tickets)

print()
print_report([])
```
